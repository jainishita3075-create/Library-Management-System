```python
import csv

with open("tips.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:

        if not row["total_bill"]:
            print("missing")

        if not row["tip"].isdigit():
            print("Invalid number")

        if not row["sex"]:
            print("missing")

        print(row)
```
