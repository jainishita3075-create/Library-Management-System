```python
import csv

with open("students.csv", "r") as file:
    reader = csv.DictReader(file)

    for student in reader:

        if not student["name"]:
            print("Name is missing")

        try:
            age = int(student["age"])
            marks = int(student["marks"])

            if age < 0:
                print("Invalid age")

            if not 0 <= marks <= 100:
                print("Invalid marks")

        except ValueError:
            print("Age and marks must be numbers")
```
